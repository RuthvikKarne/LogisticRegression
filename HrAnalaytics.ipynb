```python
#use label encoder Attrition,BusinessTravel,DailyRate,Department,EducationField,Gender,JobRole,MaritalStatus,OverTime
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data["Attrition"]=le.fit_transform(data["Attrition"])
data["BusinessTravel"]=le.fit_transform(data["BusinessTravel"])
data["Department"]=le.fit_transform(data["Department"])
data["EducationField"]=le.fit_transform(data["EducationField"])
data["JobRole"]=le.fit_transform(data["JobRole"])
data["OverTime"]=le.fit_transform(data["OverTime"])
```

```python
# Encode remaining categorical columns
from sklearn.preprocessing import LabelEncoder
for col in data.select_dtypes(include=['object']).columns:
    if col != 'Attrition':
        data[col] = LabelEncoder().fit_transform(data[col])
```

```python
# Read original data to get the string mappings
data_original = pd.read_csv("hremployees.csv")

# Create mapping dictionaries for categorical features
category_mappings = {}
categorical_cols_original = ['BusinessTravel', 'Department', 'EducationField', 'JobRole', 'Gender', 'MaritalStatus', 'OverTime']

for col in categorical_cols_original:
    unique_values = sorted(data_original[col].unique())
    category_mappings[col] = {i: val for i, val in enumerate(unique_values)}
    print(f"{col}: {category_mappings[col]}")

pickle.dump(category_mappings, open('category_mappings.pkl', 'wb'))
print("\nCategory mappings saved!")
```
